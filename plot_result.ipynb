```python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
import matplotlib

plt.rc('font', size = 14)
matplotlib.rcParams['pdf.fonttype'] = 42
matplotlib.rcParams['ps.fonttype'] = 42
```

```python
MSE_sigma = torch.load('MSE_sigma_epsilon0.3_N1000_T251.pt') ##(sigma, N, 5method, exp)
sigma_list = torch.load('sigma_list.pt')
```

```python
plt.figure()
plt.plot(sigma_list, MSE_sigma[:, 0, :].mean(axis=-1), label = r'Algorithm($u_t=0$)',color='C0', linestyle='--')
plt.plot(sigma_list, MSE_sigma[:, 1, :].mean(axis=-1), label = r'Algorithm($u_t\neq 0$)',color='C0')
plt.plot(sigma_list, MSE_sigma[:, 2, :].mean(axis=-1), label = r'Exact($u_t=0$)',color='C1', linestyle='--')
plt.plot(sigma_list, MSE_sigma[:, 3, :].mean(axis=-1), label = r'Exact($u_t\neq 0$)',color='C1')
plt.plot(sigma_list, MSE_sigma[:, 4, :].mean(axis=-1), label = 'Open-loop',color='black', linestyle='dotted')
plt.fill_between(sigma_list, MSE_sigma[:, 0, :].min(axis=-1), MSE_sigma[:, 0, :].max(axis=-1) , alpha = 0.2,color='C0')
plt.fill_between(sigma_list, MSE_sigma[:, 1, :].min(axis=-1), MSE_sigma[:, 1, :].max(axis=-1) , alpha = 0.2,color='C0')
plt.fill_between(sigma_list, MSE_sigma[:, 2, :].min(axis=-1), MSE_sigma[:, 2, :].max(axis=-1) , alpha = 0.2,color='C1')
plt.fill_between(sigma_list, MSE_sigma[:, 3, :].min(axis=-1), MSE_sigma[:, 3, :].max(axis=-1) , alpha = 0.2,color='C1')
plt.fill_between(sigma_list, MSE_sigma[:, 4, :].min(axis=-1), MSE_sigma[:, 4, :].max(axis=-1) , alpha = 0.2,color='black')
plt.xlabel(r'$\sigma$')
plt.ylabel('MSE')
plt.legend()
# plt.xscale('log')
plt.yscale('log')
plt.savefig('MSE_sigma_epsilon03_N1000_T251.pdf')
```
